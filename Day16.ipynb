```python
import pandas as pd
import numpy as np
df = pd.read_csv("Day16_Student_Wellbeing_Survey.csv")
print("Shape:", df.shape)
display(df.head())
cols = ["Weekly_Study_Hours","Average_Sleep_Hours","Daily_Screen_Time_Hours","Stress_Score","Academic_Readiness_Score"]
print("\nDESCRIPTIVE STATISTICS")
for c in cols:
    x = df[c]
    q1,q3 = x.quantile([.25,.75])
    print(f"\n{c}: Mean={x.mean():.2f}, Median={x.median():.2f}, "
          f"Mode={x.mode()[0]:.2f}, Range={x.max()-x.min():.2f}, "
          f"Var={x.var():.2f}, SD={x.std():.2f}, Q1={q1:.2f}, "
          f"Q3={q3:.2f}, IQR={q3-q1:.2f}")

sd=df[cols].std()
print("\nGreatest variability:",sd.idxmax(),f"(SD={sd.max():.2f})")

print("\nOUTLIERS (IQR METHOD)")
outcols=["Weekly_Study_Hours","Daily_Screen_Time_Hours","Commute_Time_Minutes","Monthly_Discretionary_Spending"]

for c in outcols:
    x=df[c]
    q1,q3=x.quantile([.25,.75]); iqr=q3-q1
    out=x[(x<q1-1.5*iqr)|(x>q3+1.5*iqr)]
    print(f"{c}: {len(out)} outliers | Limits=({q1-1.5*iqr:.2f}, {q3+1.5*iqr:.2f})")

c="Weekly_Study_Hours"; x=df[c]
q1,q3=x.quantile([.25,.75]); iqr=q3-q1
clean=x[(x>=q1-1.5*iqr)&(x<=q3+1.5*iqr)]
print(f"\nStudy Hours Before: Mean={x.mean():.2f}, Median={x.median():.2f}")
print(f"Study Hours After : Mean={clean.mean():.2f}, Median={clean.median():.2f}")

A=df.Part_Time_Job.eq("Yes")
B=df.Stress_Score.ge(7)
C=df.Scholarship.eq("Yes")
D=df.Exercise_Days_Per_Week.ge(3)

PA,PB,PC,PD=A.mean(),B.mean(),C.mean(),D.mean()
PAB=(A&B).mean()
PAB_or=(A|B).mean()
PAgB=PAB/PB
PBgA=PAB/PA

print("\nPROBABILITY")
print(f"P(A)={PA:.4f}, P(B)={PB:.4f}, P(C)={PC:.4f}, P(D)={PD:.4f}")
print(f"P(A or B)={PAB_or:.4f}, P(A and B)={PAB:.4f}")
print(f"P(A|B)={PAgB:.4f}, P(B|A)={PBgA:.4f}")

Y1=df.Year_of_Study.eq(1); Y4=df.Year_of_Study.eq(4)
print("\nEVENTS")
print("Year 1 & Year 4 mutually exclusive:",not (Y1&Y4).any())
print(f"P(A&B)={PAB:.4f}, P(A)*P(B)={PA*PB:.4f}")
print("Independent:",np.isclose(PAB,PA*PB,atol=.001))

PnotA=(~A).mean()
PBgnotA=(B&~A).sum()/(~A).sum()
bayes=PBgA*PA/(PBgA*PA+PBgnotA*PnotA)

print("\nBAYES' THEOREM")
print(f"P(B|A)={PBgA:.4f}, P(B|not A)={PBgnotA:.4f}")
print(f"Bayes P(A|B)={bayes:.4f}, Direct P(A|B)={PAgB:.4f}")

x=df.Academic_Readiness_Score
mu,sd=x.mean(),x.std()
hi,lo=x.max(),x.min()
zh,zl=(hi-mu)/sd,(lo-mu)/sd

print("\nNORMAL DISTRIBUTION")
print(f"Mean={mu:.2f}, SD={sd:.2f}")
print(f"Highest={hi:.2f}, Z={zh:.2f} → {zh:.2f} SD above mean")
print(f"Lowest={lo:.2f}, Z={zl:.2f} → {abs(zl):.2f} SD below mean")
print(f"68%: {mu-sd:.2f} to {mu+sd:.2f}")
print(f"95%: {mu-2*sd:.2f} to {mu+2*sd:.2f}")
print(f"99.7%: {mu-3*sd:.2f} to {mu+3*sd:.2f}")

print("\n5 OBSERVATIONS")
print(f"1. {sd:.2f} is the SD of Academic Readiness.")
print(f"2. {df[cols].std().idxmax()} shows the greatest variability.")
print(f"3. {PA:.2%} of students have a part-time job.")
print(f"4. A and B are {'independent' if np.isclose(PAB,PA*PB,atol=.001) else 'not independent'}.")
print(f"5. Academic Readiness ranges from {lo:.1f} to {hi:.1f}, with mean {mu:.2f}.")
```
